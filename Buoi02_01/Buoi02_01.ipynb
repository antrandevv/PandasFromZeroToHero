**Câu 9.** Lọc ra các đơn hàng có mức giảm giá (`Discount`) từ 0.3 trở lên. Doanh số (`SALES`) trung bình của nhóm này là bao nhiêu, so với doanh số trung bình toàn bộ dữ liệu?

```python
import pandas as pd
df=pd.read_excel('sales_data_sample.xlsx')
print(df[df['Discount']>=0.3]) #1
discountMean=df[df['Discount']>=0.3]['SALES'].mean()
overalMean=df['SALES'].mean()
print('Trung binh nhom discount >=0.3: '+str(discountMean));
print('So voi trung binh toan bo data: '+str(discountMean-overalMean));
print('Phan tram chenh lech: '+ str(((discountMean-overalMean)/overalMean)*100))
```
